```python
import random
from music21 import stream, note, metadata
from datetime import datetime
```

```python
def gen_n_notes(n):
    notes_list = ['C', 'C#', 'D', 'D#', 'E', 'F', 'F#', 'G', 'G#', 'A', 'A#', 'B']
    durations_list = [4, 2, 1, 0.5, 0.25]  # Whole, Half, Quarter, Eighth, Sixteenth

    s = stream.Stream()
    s.metadata = metadata.Metadata()
    s.metadata.title = 'Sight Reading Exercise'
    s.metadata.composer = 'nit'

    # Generate and add n random notes with random durations to the stream
    for _ in range(n):
        random_note_name = random.choice(notes_list)
        random_duration = random.choice(durations_list)
        # Create a new note with the random name and duration
        new_note = note.Note(random_note_name, quarterLength=random_duration)
        s.append(new_note)

    fp = f'./random_notes/random_notes_{datetime.now().strftime("%Y-%m-%d_%H-%M-%S")}.musicxml'
    s.write('musicxml', fp=fp)
    return fp
```
