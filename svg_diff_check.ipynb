```python
import subprocess
```

```python
def generate_diff(old_file, new_file):
    try:
        old_file_path = f'./sheet_music/{old_file}.svg'
        new_file_path = f'./sheet_music/{new_file}.svg'
        output_file = f'./diffs/{old_file}_{new_file}_diff.txt'
        
        command = f"diff {old_file_path} {new_file_path} > {output_file}"
        result = subprocess.run(command, shell=True, capture_output=True, text=True)
        
        if result.stderr:
            print(result.stderr)
        return output_file
    
    except Exception as e:
        print("Error:", e)
```

```python
def generate_diff_file(old_file_path, new_file_path, output_file):
    output = {
        'black': [],
        'red': [],
        'green': [],
    }

    with open(old_file_path, 'r') as file1, open(new_file_path, 'r') as file2:
        old_lines = file1.readlines()
        new_lines = file2.readlines()
        
        for line in old_lines:
            if '</svg>' in line:
                continue
            color = 'black' if line in new_lines else 'red'
            output[color].append(line)

        for line in new_lines:
            if line not in old_lines and '</svg>' not in line:
                output['green'].append(line)

    with open(output_file, 'w') as output_file:
        for color, lines in output.items():
            for line in lines:
                if color != 'black':
                    line = f'{line[:-5]}" style="stroke:{color}; fill:{color};"/>\n'
                output_file.write(line)
        output_file.write('</svg>')
```

```python
old_file = 'Test_Score-1'
new_file = 'Test_Score-2'
old_file_path = f'./sheet_music/{old_file}.svg'
new_file_path = f'./sheet_music/{new_file}.svg'
output_file = f'./diffs/{old_file}_{new_file}_diff.svg'
generate_diff_file(old_file_path, new_file_path, output_file)
```
